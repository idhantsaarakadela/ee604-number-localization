```python
from pathlib import Path
import os
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

PROJECT = Path(os.environ.get('EE604_PROJECT', '/content/drive/MyDrive/NumericalTextLocalization'))
OUTPUT = Path(os.environ.get('EE604_OUTPUT', str(PROJECT / 'results/M5_integration')))
OUTPUT.mkdir(parents=True, exist_ok=True)
fig, ax = plt.subplots(figsize=(11, 3.5))
ax.set(xlim=(0, 11), ylim=(0, 3.5))
ax.axis('off')
def panel(x, y, w, h, text, color='#eaf1f8'):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle='round,pad=0.08', linewidth=1,
                              edgecolor='#24465c', facecolor=color))
    ax.text(x + w / 2, y + h / 2, text, ha='center', va='center', fontsize=10)
def arrow(start, end):
    ax.annotate('', xy=end, xytext=start, arrowprops={'arrowstyle': '->', 'lw': 1.4, 'color': '#24465c'})
panel(.15, 2.15, 2.0, .9, 'SVHN image\nOne primary number')
panel(2.85, 2.15, 2.0, .9, 'YOLOv8n\nDigit classes + boxes')
panel(5.55, 2.15, 2.0, .9, 'Confidence + NMS\nRetained detections')
panel(8.25, 2.15, 2.5, .9, 'Left-to-right ordering\nDigit centres')
arrow((2.25, 2.6), (2.75, 2.6))
arrow((4.95, 2.6), (5.45, 2.6))
arrow((7.65, 2.6), (8.15, 2.6))
panel(8.25, .35, 2.5, .9, 'Concatenate labels\nNumber string', '#e8f4ec')
panel(5.55, .35, 2.0, .9, 'Enclose all digits\nSequence box', '#e8f4ec')
arrow((9.5, 2.05), (9.5, 1.35))
arrow((8.2, 2.2), (6.6, 1.35))
panel(.15, .35, 4.7, .9, 'Evaluation against ground truth\nExact text | Box IoU | Joint success', '#fff3dc')
arrow((5.45, .8), (4.95, .8))
ax.text(5.5, 3.32, 'Existing inference pipeline and report evaluation', ha='center', fontsize=12, weight='bold')
fig.savefig(OUTPUT / 'pipeline.pdf', bbox_inches='tight')
plt.close(fig)
print('Saved pipeline.pdf')
```
