## Motivation and scope

The task requires a number string and an enclosing box. SVHN supplies per-digit annotations, making digit detection a natural intermediate representation. A frozen detector lets the project study the effect of grouping while separately measuring exact recognition, localization and their conjunction.

## Literature comparison

| Study | Main representation | Relevance to this project |
|---|---|---|
| Netzer et al., 2011 | SVHN and learned digit features | Dataset and natural-image digit recognition |
| Goodfellow et al., 2014 | CNN predicts a complete digit sequence | Direct sequence recognition alternative |
| Redmon et al., 2016 | Single-stage object detection | Foundation of the detector family |
| Shi et al., 2017 | Convolutional and recurrent sequence recognition | Variable-length recognition without explicit character segmentation |
| Lin et al., 2017 | Feature pyramids | Multi-scale features useful for varied digit sizes |
| Zhou et al., 2017 | Oriented word and line localization | Region detection without digit transcription |
| Baek et al., 2019 | Character regions and affinities | Learned character grouping contrasts with spatial heuristics |

The comparisons are methodological. Published benchmark scores use different data, preprocessing and tasks, so they are not direct experimental baselines here.

## Contribution statement

The project applies YOLOv8n to SVHN digit localization and studies numerical sequence reconstruction. Its contribution is the controlled grouping comparison and the explicit separation of exact recognition, localization and joint success. The audit reconstructs metric denominators from cached predictions. The experiments support only a small localization-F1 difference between saved grouping runs and do not establish a recognition advantage for geometry-aware grouping.
